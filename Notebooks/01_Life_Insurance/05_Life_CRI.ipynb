# Life Claims Intelligence — Claims Reliability Index (CRI)

## Objective

The Claims Reliability Index (CRI) is an independent analytical framework
developed in this project to evaluate insurer reliability beyond the
headline Claim Settlement Ratio (CSR).

The index combines multiple claims-reliability dimensions into a common
score so that insurers can be compared using a broader view of claim
settlement performance.

> **Important:** CRI is an independent analytical index developed for this
> project. It is not an IRDAI-official score or regulatory rating.

## Why CRI?

A high count-based Claim Settlement Ratio does not necessarily imply that
an insurer settles a similarly high proportion of claim value.

Therefore, the CRI considers value-based settlement performance,
count–value consistency, and claim outcome/processing reliability rather
than relying on a single headline metric.

## Life-specific Adaptation

The original project framework proposed four CRI components. During actual
Life Insurance data analysis, Complaint Rate and Claim Value Disparity could
not be retained as reliable Life-level components.

Therefore, the Life CRI will be adapted to the metrics that are actually
available and validated in the Life dataset.

The methodology will be documented transparently rather than introducing
unsupported or fabricated variables.

## Final Life CRI Components

Based on the structure and availability of the Life Insurance dataset,
the CRI will use three primary components:

1. **Value Reliability**
   - Value_CSR
   - Higher values indicate better value-based claim settlement.

2. **Count–Value Consistency**
   - Count_Value_Gap
   - Lower values indicate greater consistency between count-based and
     value-based settlement performance.

3. **Outcome Reliability**
   - Repudiation_Rate
   - Pending_Rate
   - Unclaimed_Ratio
   - These metrics will be combined into a single outcome-reliability
     component because they represent related claim-outcome dimensions.

### Supporting Metrics

Count_CSR and Long_Pending_Share will be retained as supporting
diagnostic metrics. They will not receive independent CRI weights in order
to reduce redundancy and avoid giving excessive weight to closely related
claim-status measures.

### Baseline Weighting

The three primary CRI components will initially receive equal weights:

- Value Reliability: 33.33%
- Count–Value Consistency: 33.33%
- Outcome Reliability: 33.33%

These baseline weights will later be subjected to sensitivity analysis.

### 1. Creating Value_Reliability_Score

```python
cri_df["Value_Reliability_Score"] = (
    (cri_df["Value_CSR"] - cri_df["Value_CSR"].min())
    /
    (cri_df["Value_CSR"].max() - cri_df["Value_CSR"].min())
)
```

```python
cri_df[
    ["Value_CSR", "Value_Reliability_Score"]
].head(10)
```

```python
cri_df["Value_Reliability_Score"].describe()
```

---

### 2.Creating Count_Value_Consistency_Score

```python
cri_df["Absolute_Count_Value_Gap"] = cri_df["Count_Value_Gap"].abs()
```

```python
cri_df["Count_Value_Consistency_Score"] = (
    (
        cri_df["Absolute_Count_Value_Gap"].max()
        - cri_df["Absolute_Count_Value_Gap"]
    )
    /
    (
        cri_df["Absolute_Count_Value_Gap"].max()
        - cri_df["Absolute_Count_Value_Gap"].min()
    )
)
```

```python
cri_df[
    [
        "Count_Value_Gap",
        "Absolute_Count_Value_Gap",
        "Count_Value_Consistency_Score"
    ]
].head(10)
```

----

### 3. Pending Rate

```python
cri_df["Pending_Score"] = (
    (cri_df["Pending_Rate"].max() - cri_df["Pending_Rate"])
    /
    (cri_df["Pending_Rate"].max() - cri_df["Pending_Rate"].min())
)
```

### 4. Unclaimed Ratio

```python
cri_df["Unclaimed_Score"] = (
    (cri_df["Unclaimed_Ratio"].max() - cri_df["Unclaimed_Ratio"])
    /
    (cri_df["Unclaimed_Ratio"].max() - cri_df["Unclaimed_Ratio"].min())
)
```

### 5. Repudiation Rate

```python
cri_df["Repudiation_Score"] = (
    (cri_df["Repudiation_Rate"].max() - cri_df["Repudiation_Rate"])
    /
    (cri_df["Repudiation_Rate"].max() - cri_df["Repudiation_Rate"].min())
)
```

---

### Outcome Reliability

```python
cri_df["Outcome_Reliability_Score"] = (
    cri_df[
        [
            "Repudiation_Score",
            "Pending_Score",
            "Unclaimed_Score"
        ]
    ].mean(axis=1)
)
```

## CRI Methodology — Final

The Life Claims Reliability Index (CRI) is an independent analytical index
developed in this project to summarize insurer reliability across multiple
claims-performance dimensions.

It is not an IRDAI-official score or regulatory rating.

### CRI Components

The Life-specific CRI uses three primary components:

1. **Value Reliability**
   - Based on Value_CSR.
   - Higher Value_CSR indicates better value-based claim settlement
     performance.

2. **Count–Value Consistency**
   - Based on the absolute Count–Value Gap.
   - The gap is defined as Count_CSR − Value_CSR.
   - Because consistency is about the distance between the two measures,
     the absolute gap is used.
   - Smaller absolute gaps receive higher consistency scores.

3. **Outcome Reliability**
   - Combines Repudiation_Rate, Pending_Rate, and Unclaimed_Ratio.
   - Each metric is normalized separately and directionally aligned so
     that higher normalized scores indicate better reliability.
   - The three normalized scores are averaged to create the
     Outcome_Reliability_Score.

### Normalization

Min-max normalization was used to place metrics on a common 0–1 scale.

For higher-is-better metrics:

    (x − min(x)) / (max(x) − min(x))

For lower-is-better metrics:

    (max(x) − x) / (max(x) − min(x))

### Baseline Weighting

The three CRI components receive equal baseline weights:

- Value Reliability: 33.33%
- Count–Value Consistency: 33.33%
- Outcome Reliability: 33.33%

The final CRI is calculated on a 0–100 scale.

### Data Availability

Six insurer-year observations did not contain the core metrics required for
CRI calculation. These observations were excluded from the CRI working
dataset but retained in the master dataset.

### Sensitivity Analysis

The baseline weighting was tested against three alternative scenarios:

- Value-focused: 40% / 30% / 30%
- Consistency-focused: 30% / 40% / 30%
- Outcome-focused: 30% / 30% / 40%

Spearman correlations between the baseline and alternative rankings were
approximately 0.999, indicating that the insurer-year ranking was highly
stable under these moderate weight changes.

### Sanity Check

High-CRI observations generally showed strong performance across the three
CRI components, while low-CRI observations showed weakness in one or more
major components.

This confirms that the index behaves directionally as intended. The sanity
check is not treated as independent statistical validation.

# CRI — Conclusion

The Life Claims Reliability Index (CRI) was developed to provide a consolidated
view of insurer-level claims reliability across multiple performance dimensions.

The final CRI combines three components:

- **Value Reliability** — based on Value_CSR
- **Count–Value Consistency** — based on the absolute Count–Value Gap
- **Outcome Reliability** — based on Repudiation Rate, Pending Rate, and
  Unclaimed Ratio

All component metrics were normalized to a common 0–1 scale and combined using
equal baseline weights. The final CRI is reported on a 0–100 scale.

A total of **116 insurer-year observations** were available for CRI calculation.
Six insurer-year observations were excluded from the CRI working dataset because
the core metrics required for CRI calculation were unavailable. These
observations remain retained in the master dataset.

### Sensitivity Analysis

The baseline equal-weight CRI was tested against alternative weighting
scenarios. The resulting ranking correlations were approximately **0.999**
across the tested scenarios, indicating that the insurer-year ranking was highly
stable under moderate changes in component weights.

### Sanity Check

The highest-CRI observations generally showed strong performance across the
three CRI components, while lower-CRI observations showed weakness in one or
more components. This indicates that the CRI behaves directionally as intended.

The CRI is an **independent analytical index developed for this project** and
should not be interpreted as an official regulatory or IRDAI rating.
